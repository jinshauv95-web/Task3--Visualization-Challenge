#  📘Visualization Challenge

## 🎯 Objective

The objective of this task is to create professional visualizations that communicate business insights from the cleaned dataset. The focus is on clarity, correct chart selection, and meaningful interpretation.

## 📂 Dataset

cleaned_ecommerce_data.csv

## 🔷 1. Import Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

## 🔷 2. Load Dataset

```python
df = pd.read_csv("cleaned_ecommerce_data.csv")

# Convert date
df['order_date'] = pd.to_datetime(df['order_date'])
df['month'] = df['order_date'].dt.to_period('M')
```

## 📊 BAR CHARTS

### 🔷 1. Revenue by Category

```python
category_sales = df.groupby('_category')['total'].sum().sort_values(ascending=False)

plt.figure()
category_sales.plot(kind='bar')
plt.title("Revenue by Category")
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.show()
```

* Identifies which product categories generate the highest revenue

### 🔷 2. Top Products by Units Sold

```python
top_products = df.groupby('product')['quantity'].sum().sort_values(ascending=False).head(10)

plt.figure()
top_products.plot(kind='bar')
plt.title("Top 10 Products by Units Sold")
plt.xlabel("Product")
plt.ylabel("Units Sold")
plt.xticks(rotation=45)
plt.show()
```

* Identifies best-selling products

### 🔷 3. Order Status Distribution

```python
status_counts = df['status'].value_counts()

plt.figure()
status_counts.plot(kind='bar')
plt.title("Order Status Distribution")
plt.xlabel("Status")
plt.ylabel("Count")
plt.show()
```

* Helps evaluate delivery efficiency

## 📈 LINE CHARTS

### 🔷 4. Monthly Sales Trend

```python
monthly_sales = df.groupby('month')['total'].sum()

plt.figure()
monthly_sales.plot(marker='o')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

* Shows growth and seasonal demand patterns

### 🔷 5. Weekly Order Trend

```python
df['week'] = df['order_date'].dt.to_period('W')

weekly_orders = df.groupby('week')['order_id'].count()

plt.figure()
weekly_orders.plot(marker='o')
plt.title("Weekly Order Trend")
plt.xlabel("Week")
plt.ylabel("Orders")
plt.show()
```

* Helps understand short-term demand changes

### 🔷 6. Category Revenue Trend Over Time

```python
category_trend = df.groupby(['month', '_category'])['total'].sum().unstack()

plt.figure()
category_trend.plot()
plt.title("Category Revenue Trend Over Time")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

* Shows how each category performs over time

## 🥧 PIE CHARTS

### 🔷 7. Sales by Payment Method

```python
payment_sales = df.groupby('payment_method')['total'].sum()

plt.figure()
payment_sales.plot(kind='pie', autopct='%1.1f%%')
plt.title("Sales by Payment Method")
plt.ylabel("")
plt.show()
```

* Shows customer payment preferences

### 🔷 8. Order Status Share

```python
status_share = df['status'].value_counts()

plt.figure()
status_share.plot(kind='pie', autopct='%1.1f%%')
plt.title("Order Status Share")
plt.ylabel("")
plt.show()
```

* Shows proportion of completed vs pending orders

## 📊 DASHBOARD MOCKUP

### 🔷 9. Combined Dashboard View

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# KPI - Revenue
axes[0,0].text(0.5, 0.5, f"Revenue\n{df['total'].sum():,.0f}", ha='center')
axes[0,0].set_title("Total Revenue")
axes[0,0].axis('off')

# Category Sales
category_sales.plot(kind='bar', ax=axes[0,1])
axes[0,1].set_title("Category Sales")

# Monthly Trend
monthly_sales.plot(ax=axes[1,0])
axes[1,0].set_title("Monthly Sales Trend")

# Payment Method
payment_sales.plot(kind='bar', ax=axes[1,1])
axes[1,1].set_title("Payment Method")

plt.tight_layout()
plt.show()
```

## 📝 FINAL OBSERVATIONS
### Key Findings

1. Product categories significantly influence revenue distribution.
2. A few products dominate total units sold.
3. Sales show clear monthly variations indicating seasonal patterns.
4. Payment methods reflect customer preferences.
5. Order status distribution helps evaluate operational efficiency.

The visualizations provide clear insights into sales performance,
